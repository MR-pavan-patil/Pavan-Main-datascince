```python
import tabula
import pandas as pd

pdf_path = "clean .pdf"   # your PDF file
output_csv = "cleaned.csv"

# Read all tables from all pages
tables = tabula.read_pdf(
    pdf_path,
    pages="all",
    multiple_tables=True
)

# Combine all tables
df = pd.concat(tables, ignore_index=True)

# Clean column names
df.columns = [str(c).strip().replace("\n", "_") for c in df.columns]

# Save to single CSV
df.to_csv(output_csv, index=False)

print("✅ Full single CSV generated:", output_csv)
```
